# Data cleaning/ Data Wrangling

```python
unique_ids = bank_1['CustomerId'].nunique()
print(f"Total rows: {len(bank_1)} | Unique Customer IDs: {unique_ids}")
```

```python
bank_clean = bank_1.drop(columns=['RowNumber', 'Surname']).copy()
```

```python
bank_clean.columns = (
    bank_clean.columns
    .str.strip()
    .str.replace(r'([a-z])([A-Z])', r'\1_\2', regex=True)
    .str.lower()
)
```

```python
bank_clean.columns
```

```python
bank_clean = bank_clean.rename(columns={
    'has_cr_card': 'has_credit_card',
    'is_active_member': 'is_active'
})
```

```python
bank_clean.columns
```

# Exporting Clean Data

```python
processed_dir = r"D:\Project\Banking_Customer_churn&_Retention\Data\processed"
os.makedirs(processed_dir, exist_ok=True)
```

```python
clean_file_path = os.path.join(processed_dir, "bank_churn_clean.csv")
bank_clean.to_csv(clean_file_path, index=False)
```
